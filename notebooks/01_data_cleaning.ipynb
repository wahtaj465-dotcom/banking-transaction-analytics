```python
# Validate promary keys - Banking dataset cannot have duplicate ids
primary_keys = {
    "Customers": ("customer_id", customers),
    "Accounts": ("account_id", accounts),
    "Transactions": ("transaction_id", transactions),
    "Merchants": ("merchant_id", merchants),
    "Branches": ("branch_id", branches)
}

pk_report = []

for table, (pk, df) in primary_keys.items():

    pk_report.append({
        "Table": table,
        "Primary Key": pk,
        "Duplicates": df[pk].duplicated().sum(),
        "Missing": df[pk].isnull().sum()
    })

pd.DataFrame(pk_report)
```

```python
#validate foreign keys
invalid_accounts = transactions[
    ~transactions["account_id"].isin(accounts["account_id"])
]

print("Invalid Account IDs :", len(invalid_accounts))
```

```python
''' same as left join
SELECT *
FROM transactions t
LEFT JOIN accounts a
ON t.account_id = a.account_id
WHERE a.account_id IS NULL;
'''


invalid_merchants = transactions[
    ~transactions["merchant_id"].isin(merchants["merchant_id"])
]

invalid_branches = transactions[
    ~transactions["branch_id"].isin(branches["branch_id"])
]

print("Invalid Merchant IDs :", len(invalid_merchants))
print("Invalid Branch IDs :", len(invalid_branches))
```

## Data Integration

We combine all relational tables into one analytical dataset using LEFT JOIN.

```python
master = transactions.merge(
    accounts,
    on="account_id",
    how="left"
)

master.head()
```

```python
master = master.merge(
    customers,
    on="customer_id",
    how="left"
)

master.head()
```

```python
master = master.merge(
    merchants,
    on="merchant_id",
    how="left"
)

master.head()
```

```python
master = master.merge(
    branches,
    on="branch_id",
    how="left"
)

master.head()
```

Feature Engineering


```python
# Date Features

master["year"] = master["transaction_datetime"].dt.year

master["month"] = master["transaction_datetime"].dt.month_name()

master["month_no"] = master["transaction_datetime"].dt.month

master["quarter"] = "Q" + master["transaction_datetime"].dt.quarter.astype(str)

master["day"] = master["transaction_datetime"].dt.day

master["weekday"] = master["transaction_datetime"].dt.day_name()

master["hour"] = master["transaction_datetime"].dt.hour
```

```python
master[
    ["transaction_datetime",
     "year",
     "month",
     "quarter",
     "weekday",
     "hour"]
].head()
```

```python
# Create a weekend flag as banks check for weekend or weekdays
```

```python
master["is_weekend"] = master["weekday"].isin([
    "Saturday",
    "Sunday"
])
```

```python
master[["weekday", "is_weekend"]].head()
```

```python
# Revenue classification column
master["transaction_band"] = pd.cut(
    master["amount"],
    bins=[0,1000,5000,10000,50000,np.inf],
    labels=[
        "Low",
        "Medium",
        "High",
        "Very High",
        "Premium"
    ]
)
```
