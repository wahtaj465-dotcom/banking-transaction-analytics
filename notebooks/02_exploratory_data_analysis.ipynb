```python
import micropip
await micropip.install("plotly")
```

```python
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go

pd.set_option("display.max_columns", None)

print("Libraries imported successfully!")
```

```python
master = pd.read_csv("banking_master.csv")

master["transaction_datetime"] = pd.to_datetime(master["transaction_datetime"])

master.head()
```

```python
#Executive Kpi Calculation that appears on the top of the dashboard

total_revenue = master["amount"].sum()

total_transactions = master["transaction_id"].count()

unique_customers = master["customer_id"].nunique()

avg_transaction = master["amount"].mean()

success_rate = (
    master["transaction_status"]
    .eq("Success")
    .mean()*100
)
```

```python
# Converting Kpi to dataframe for better reading
kpi = pd.DataFrame({
    "KPI":[
        "Total Revenue",
        "Transactions",
        "Customers",
        "Average Transaction",
        "Success Rate"
    ],
    "Value":[
        f"₹ {total_revenue:,.2f}",
        f"{total_transactions:,}",
        f"{unique_customers:,}",
        f"₹ {avg_transaction:,.2f}",
        f"{success_rate:.2f}%"
    ]
})

kpi
```

```python
#Monthly Revenue Trend
monthly = (
    master
    .groupby(["year","month_no","month"], as_index=False)
    ["amount"]
    .sum()
    .sort_values(["year","month_no"])
)

monthly
```

```python
fig = px.line(
    monthly,
    x="month",
    y="amount",
    color="year",
    markers=True,
    title="Monthly Revenue Trend"
)

fig.update_layout(
    xaxis_title="Month",
    yaxis_title="Revenue"
)

from IPython.display import display, HTML

display(
    HTML(
        fig.to_html(
            include_plotlyjs="cdn",
            full_html=False,
            default_width="900px",
            default_height="500px"
        )
    )
)
```

```python
# Revenue by region
region = (
    master
    .groupby("region", as_index=False)
    ["amount"]
    .sum()
    .sort_values("amount", ascending=False)
)

region
```

```python
from IPython.display import HTML, display

def show_plot(fig, width=900, height=500):
    display(
        HTML(
            fig.to_html(
                include_plotlyjs="cdn",
                full_html=False,
                default_width=f"{width}px",
                default_height=f"{height}px"
            )
        )
    )

fig = px.pie(
    region,
    names="region",
    values="amount",
    hole=0.45,
    title="Regional Revenue Distribution"
)

show_plot(fig)


```

```python
#Merchant By Category
category = (
    master
    .groupby("category", as_index=False)
    ["amount"]
    .sum()
    .sort_values("amount", ascending=False)
)

category
```

```python
fig = px.bar(
    category,
    x="amount",
    y="category",
    orientation="h",
    text_auto=".2s",
    title="Revenue by Merchant Category",
    color="amount",
    color_continuous_scale="Blues"
)

fig.update_layout(
    xaxis_title="Revenue (₹)",
    yaxis_title="Merchant Category",
    template="plotly_white"
)

show_plot(fig)
```

```python
#Branch Performance Analysis

branch = (
    master
    .groupby("branch_name", as_index=False)["amount"]
    .sum()
    .sort_values("amount", ascending=False)
    .head(10)
)

branch
```

```python
fig = px.bar(
    branch,
    x="amount",
    y="branch_name",
    orientation="h",
    text_auto=".2s",
    title="Top 10 Revenue Generating Branches",
    color="amount",
    color_continuous_scale="Blues"
)

fig.update_layout(
    xaxis_title="Revenue (₹)",
    yaxis_title="Branch"
)

show_plot(fig)
```

```python
# Payment Method Analysis
payment = (
    master
    .groupby("payment_method", as_index=False)
    .agg(
        Transactions=("transaction_id", "count"),
        Revenue=("amount", "sum")
    )
    .sort_values("Transactions", ascending=False)
)

payment
```

```python
fig = px.bar(
    payment,
    x="payment_method",
    y="Transactions",
    text_auto=True,
    title="Transaction Volume by Payment Method",
    color="Transactions",
    color_continuous_scale="Greens"
)

show_plot(fig)
```

```python
#Banking Channel Analysis
channel = (
    master
    .groupby("channel", as_index=False)["amount"]
    .sum()
    .sort_values("amount", ascending=False)
)

channel

```

```python
fig = px.pie(
    channel,
    names="channel",
    values="amount",
    hole=0.45,
    title="Revenue Distribution by Banking Channel"
)

show_plot(fig)
```

```python
#Customer Age Distribution
fig = px.histogram(
    master,
    x="age",
    nbins=20,
    title="Customer Age Distribution",
    color_discrete_sequence=["#2563EB"]
)

fig.update_layout(
    xaxis_title="Age",
    yaxis_title="Number of Customers"
)

show_plot(fig)
```

```python
# Occupation wise spending
occupation = (
    master
    .groupby("occupation", as_index=False)["amount"]
    .sum()
    .sort_values("amount", ascending=False)
)

occupation
```

```python
fig = px.bar(
    occupation,
    x="occupation",
    y="amount",
    text_auto=".2s",
    title="Revenue by Customer Occupation",
    color="amount",
    color_continuous_scale="Purples"
)

fig.update_layout(xaxis_title="Occupation")

show_plot(fig)
```

```python
#Gender Spending Comparison
gender = (
    master
    .groupby("gender", as_index=False)
    .agg(
        Revenue=("amount", "sum"),
        Customers=("customer_id", "nunique")
    )
)

gender
```

```python
fig = px.bar(
    gender,
    x="gender",
    y="Revenue",
    text_auto=".2s",
    color="gender",
    title="Revenue by Gender"
)

show_plot(fig)
```

## Correlation Analysis

Analyze the relationship between balance, income, age and transaction amount.

```python
numeric = master[
    ["amount",
     "opening_balance",
     "closing_balance",
     "balance",
     "annual_income",
     "age"]
]

corr = numeric.corr(numeric_only=True).round(2)

corr
```

```python
fig = px.imshow(
    corr,
    text_auto=True,
    color_continuous_scale="RdBu",
    title="Correlation Heatmap"
)

show_plot(fig)
```

```python
# Peak Transaction Hours
hourly = (
    master
    .groupby("hour", as_index=False)
    .agg(
        Transactions=("transaction_id", "count"),
        Revenue=("amount", "sum")
    )
)

hourly
```

```python
fig = px.line(
    hourly,
    x="hour",
    y="Transactions",
    markers=True,
    title="Peak Transaction Hours"
)

fig.update_layout(
    xaxis=dict(dtick=1),
    xaxis_title="Hour of Day",
    yaxis_title="Number of Transactions"
)

show_plot(fig)
```

```python
#Weekend Vs Weekday
weekend = (
    master
    .groupby("is_weekend", as_index=False)
    .agg(
        Revenue=("amount", "sum"),
        Transactions=("transaction_id", "count")
    )
)

weekend["Day Type"] = weekend["is_weekend"].map({
    True:"Weekend",
    False:"Weekday"
})

weekend
```

```python
fig = px.bar(
    weekend,
    x="Day Type",
    y="Revenue",
    text_auto=".2s",
    color="Day Type",
    title="Weekend vs Weekday Revenue"
)

show_plot(fig)
```

```python
#Income vs Spending
customer_spending = (
    master
    .groupby(
        ["customer_id","annual_income","gender"],
        as_index=False
    )
    .agg(
        Total_Spent=("amount","sum")
    )
)

customer_spending.head()
```

```python
fig = px.scatter(
    customer_spending,
    x="annual_income",
    y="Total_Spent",
    color="gender",
    opacity=0.6,
    title="Annual Income vs Customer Spending"
)
import numpy as np

x = customer_spending["annual_income"]
y = customer_spending["Total_Spent"]

m, b = np.polyfit(x, y, 1)

fig.add_scatter(
    x=x,
    y=m*x + b,
    mode="lines",
    name="Regression Line",
    line=dict(color="red", width=3)
)

show_plot(fig)
```

```python
#Successful Vs Failed Transactions
status = (
    master
    .groupby("transaction_status", as_index=False)
    .agg(
        Count=("transaction_id","count")
    )
)

status["Percentage"] = (
    status["Count"]/status["Count"].sum()*100
).round(2)

status
```

```python
fig = px.pie(
    status,
    names="transaction_status",
    values="Count",
    hole=0.55,
    title="Transaction Success Rate"
)

show_plot(fig)
```

```python
# Revenue by Quarter
quarter = (
    master
    .groupby("quarter", as_index=False)
    ["amount"]
    .sum()
)

quarter
```

```python
fig = px.bar(
    quarter,
    x="quarter",
    y="amount",
    text_auto=".2s",
    color="amount",
    color_continuous_scale="Teal",
    title="Quarter-wise Revenue"
)

show_plot(fig)
```
